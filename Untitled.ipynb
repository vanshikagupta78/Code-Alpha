```python
from sklearn.datasets import load_iris
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
iris=load_iris()
```

```python
df=pd.DataFrame(data=iris.data,columns=iris.feature_names)
```

```python
df['species']=iris.target
```

```python
print(df.head())
```

```python
df['species']=df['species'].map({
    0:'setosa',
    1:'versicolor',
    2:'virginica'
})
print(df.head())
```

```python
g=sns.pairplot(df,hue='species',palette='Set3')
g.fig.suptitle("Iris Dataset Pairplot",y=1.02)
plt.show()
```

```python
corr=df.corr(numeric_only=True)
sns.heatmap(corr,annot=True,cmap='Blues')
plt.title("Both Length And Width")
plt.show()
```

```python
sns.scatterplot(x='sepal length (cm)',y='petal length (cm)',hue='species',data=df)
plt.title("Species")
plt.show()
```

In this task, I loaded the Iris dataset using Scikit-learn and created multiple visualizations such as pairplots,histograms,boxplots,scatter plots, and heatmaps using Matplotlib and Seaborn. These visualizations helped in understanding feature relationships, identifying patterns and deriving insights for better decision_making.
